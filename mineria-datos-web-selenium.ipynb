```python
# ==============================================================================
# PASO 0: Configuración del Entorno de Selenium Headless
# ==============================================================================
import time
import re
import pandas as pd
from bs4 import BeautifulSoup

# Importación de componentes clave de Selenium WebDriver
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def iniciar_driver_headless():
    """
    Inicializa una instancia de Chrome WebDriver configurada en modo Headless
    optimizada para entornos Linux virtualizados (Google Colab / Jupyter).
    """
    # options = webdriver.ChromeOptions()
    # options.add_argument('--headless')                  # Modo sin interfaz gráfica
    # options.add_argument('--no-sandbox')               # Requerido para entornos de contenedores
    # options.add_argument('--disable-dev-shm-usage')    # Optimización de memoria compartida
    # options.add_argument('--disable-gpu')               # Desactiva aceleración por GPU
    
    # En entornos Linux / Google Colab la ruta predeterminada del ejecutable es:
    # service = Service('/usr/bin/chromedriver')
    # driver = webdriver.Chrome(service=service, options=options)
    # driver.set_window_size(1920, 1080)

    options = webdriver.ChromeOptions()
    options.add_argument('--headless=new')
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')
    options.add_argument('--disable-gpu')

    driver = webdriver.Chrome(options=options)
    driver.set_window_size(1920, 1080)

    return driver

print("✅ Entorno de Selenium Headless configurado exitosamente.")
```

### Navegación Automatizada y Bucle de Desplazamiento Vertical (Scroll Loop)

```python
# ==============================================================================
# PASO 1: Apertura de la URL y Ejecución del Bucle de Scroll
# ==============================================================================
URL_CATALOGO = "https://www.lagranbodega.com.mx/frutas-y-verduras"

# Inicializar el navegador en segundo plano
driver = iniciar_driver_headless()
print(f"🚀 Conectando a la plataforma objetivo: {URL_CATALOGO}")

try:
    driver.get(URL_CATALOGO)
    time.sleep(4)  # Espera inicial para el renderizado del documento base

    # Parámetros del desplazamiento vertical
    MAX_SCROLLS = 8              # Número de iteraciones de desplazamiento
    PAUSA_AJAX = 2.5             # Segundos de espera para la carga de nuevos nodos
    PIXELES_DESPLAZAMIENTO = 1200 # Píxeles a desplazar por cada ciclo

    print("\n🔄 Iniciando secuencia de desplazamiento vertical (Scroll Loop)...")

    for i in range(1, MAX_SCROLLS + 1):
        # Ejecución de script JavaScript en el navegador simulado
        driver.execute_script(f"window.scrollBy(0, {PIXELES_DESPLAZAMIENTO});")
        print(f"   ── Iteración #{i}/{MAX_SCROLLS}: Scroll ejecutado. Esperando respuesta del servidor...")
        time.sleep(PAUSA_AJAX)
        
        # Conteo en tiempo real de nodos de producto presentes en el DOM
        tarjetas_detectadas = driver.find_elements(By.CSS_SELECTOR, "div.product-item, div.product-card, div.vtex-search-result-3-x-galleryItem")
        print(f"      [DOM Check] Nodos de productos detectados actualmente: {len(tarjetas_detectadas)}")

    # Captura del código HTML fuente totalmente renderizado y expandido
    html_fuente_renderizado = driver.page_source
    print("\n✅ Extracción del HTML renderizado finalizada correctamente.")

finally:
    driver.quit()
    print("🔒 Instancia de Selenium cerrada de forma segura.")
```

### Paso 2: Parseo Quirúrgico del DOM con BeautifulSoup

```python
eliminar_anuncios = True #@param {type:"boolean"}
eliminar_headers_footers = True #@param {type:"boolean"}
eliminar_sidebars = True #@param {type:"boolean"}
eliminar_metas = True #@param {type:"boolean"}

# Inspección del primer artículo en el corpus
soup_demo = BeautifulSoup(html_fuente_renderizado, 'html.parser')

print("=== ESTADO ORIGINAL DEL HTML (Primeros 500 caracteres) ===")
print(soup_demo.prettify()[:500])
print("\n" + "="*60 + "\n")

# Aplicar limpieza de elementos no deseados (Boilerplate Removal)
elementos_a_remover = []
if eliminar_anuncios:
    elementos_a_remover.extend(['.banner-ads', '.in-text-ad'])
if eliminar_headers_footers:
    elementos_a_remover.extend(['header', 'footer'])
if eliminar_sidebars:
    elementos_a_remover.extend(['.sidebar'])
if eliminar_metas:
    elementos_a_remover.extend(['head', 'aside', 'script', 'style', 'meta', 'link'])


for selector in elementos_a_remover:
    for elemento in soup_demo.select(selector):
        elemento.decompose() # Elimina el nodo del árbol DOM

print("=== ESTADO TRAS BOILERPLATE REMOVAL (Sin ruido web) ===")
print(soup_demo.prettify())
```

### Paso 3: Limpieza de Datos, Expresiones Regulares y Conversión Tabular en Pandas

```python
registros_extraidos = []

articulos = soup_demo.find_all('article')
print(f"Se encontraron {len(articulos)} articulos")
for idx, articulo in enumerate(articulos):
  if articulo:
    registro = {"id_articulo": f"{100+idx}"} # Agregar un identificador de un articulo
    # --- INSTRUCCIONES FALTANTES ---

    registro["nombre"] = articulo.find('h6').a.text.strip()  # Extrae el nombre del artículo
    precio_tag = articulo.find('span', class_='newPrice')
    # Si precio_tag existe, extraemos el texto; si no, asignamos un valor por defecto
    registro["precio"] = precio_tag.text.strip() if precio_tag else "No disponible"
    registro["imagen"] = articulo.find('div', class_='article__image').find('img')['src']  # Extrae la URL de la imagen

  registros_extraidos.append(registro) # Agregando una nuevo producto


print(registros_extraidos)
```

```python
df_productos = pd.DataFrame(registros_extraidos)
print("\n=== DataFrame de Productos Extraídos ===")
print(df_productos)
```

```python
pd.DataFrame.to_csv(df_productos, 'productos_extraidos.csv', index=False)
```
